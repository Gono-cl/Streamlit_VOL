```python
from spm_pump import SPMPump

pump = SPMPump(syringe_volume=5000)
```

```python
methilene_blue = 1
water = 2
outlet = 3 

stock_concentration = 200 # mg/L
final_volume = 5000 # ul

standard1 = 10 #mg/L
standard2 = 20 #mg/L
standard3 = 30 #mg/L
standard4 = 40 #mg/L
standard5 = 50 #mg/L

volume_standar1 = standard1 * final_volume / stock_concentration
volume_standar2 = standard2 * final_volume / stock_concentration
volume_standar3 = standard3 * final_volume / stock_concentration
volume_standar4 = standard4 * final_volume / stock_concentration
volume_standar5 = standard5 * final_volume / stock_concentration


sample1_concentration = 15 # mg/L
sample2_concentration = 25 # mg/L
sample3_concentration = 35 # mg/L

volume_sample1 = sample1_concentration * final_volume / stock_concentration
volume_sample2 = sample2_concentration * final_volume / stock_concentration
volume_sample3 = sample3_concentration * final_volume / stock_concentration


# methylene blue
pump.move_valve(methilene_blue)
pump.set_flow_rate(1000)
pump.pickup(volume_standar5)

# water
pump.move_valve(water)
pump.set_flow_rate(25000)
pump.pickup(final_volume - volume_standar5)

# outlet toward UV/Vis
pump.move_valve(outlet)
pump.set_flow_rate(1000)
pump.dispense(5000)

pump.move_valve(water)
pump.set_flow_rate(20000)
pump.pickup(5000)

pump.move_valve(outlet)
pump.set_flow_rate(20000)
pump.dispense(5000)
```
